```python
HG_df = pd.read_csv ("the 10th Hunger Games.csv")
```

```python
fig, ax = plt.subplots()
ax.scatter(HG_df["district"], HG_df["death order | books"], s=12, alpha=0.6, color='gold')
ax.set_title("District vs Death Order in Books")
ax.set_xlabel("District")
ax.set_ylabel("Death Order")
```

```python
fig, ax = plt.subplots()
ax.scatter(HG_df["district"], HG_df["death order | film"], s=12, alpha=0.6, color='gold')
ax.set_title("District vs Death Order in Film")
ax.set_xlabel("District")
ax.set_ylabel("Death Order")
```
